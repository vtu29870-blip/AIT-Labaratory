```python
def monkey_banana():
    state = ["floor", "window", False]
    print("Initial State:", state)
    print("Monkey moves to box")
    state[0] = "window"
    print("Monkey pushes box to center")
    state[1] = "center"
    print("Monkey climbs box")
    state[2] = True
    if state[2] and state[1] == "center":
        print("Monkey touches bananas!")
        print("Goal Reached!")
monkey_banana()
```
